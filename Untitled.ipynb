```python
def clean_data(df):
    df = df.copy()

    none_cols = ['PoolQC','MiscFeature','Alley','Fence','FireplaceQu','GarageType','GarageFinish',
                 'GarageQual','GarageCond','BsmtQual','BsmtCond','BsmtExposure','BsmtFinType1',
                 'BsmtFinType2','MasVnrType']
    for c in none_cols:
        df[c] = df[c].fillna('None')

    df['LotFrontage'] = df.groupby('Neighborhood')['LotFrontage'].transform(lambda x: x.fillna(x.median()))

    zero_cols = ['GarageYrBlt','MasVnrArea','GarageArea','GarageCars',
                 'BsmtFinSF1','BsmtFinSF2','BsmtUnfSF','TotalBsmtSF',
                 'BsmtFullBath','BsmtHalfBath']
    for c in zero_cols:
        if c in df.columns:
            df[c] = df[c].fillna(0)

    mode_cols = ['Electrical','MSZoning','Utilities','Exterior1st','Exterior2nd',
                 'KitchenQual','Functional','SaleType']
    for c in mode_cols:
        if c in df.columns and df[c].isnull().any():
            df[c] = df[c].fillna(df[c].mode()[0])

    df['MSSubClass'] = df['MSSubClass'].astype(str)

    qual_map = {'None':0,'Po':1,'Fa':2,'TA':3,'Gd':4,'Ex':5}
    qual_cols = ['ExterQual','ExterCond','BsmtQual','BsmtCond','HeatingQC','KitchenQual',
                 'FireplaceQu','GarageQual','GarageCond','PoolQC']
    for c in qual_cols:
        df[c] = df[c].map(qual_map)

    return df
```

```python
train_df = pd.read_csv('train.csv')
train_df = train_df[(train_df['GrLivArea']<4000) | (train_df['SalePrice']>=300000)]

train_clean = clean_data(train_df)
test_clean = clean_data(test_df)

print("train_clean shape:", train_clean.shape)
print("test_clean shape:", test_clean.shape)
```

```python
# ── 1. Mark which rows are train vs test, before combining ──
test_ids = test_clean['Id']
train_clean['is_train'] = 1
test_clean['is_train'] = 0
test_clean['SalePrice'] = np.nan   # test has no SalePrice — add a placeholder so shapes match

# ── 2. Stack them into one table and encode together ──
combined = pd.concat([train_clean, test_clean], axis=0, ignore_index=True)
combined_encoded = pd.get_dummies(combined, drop_first=True)

bool_cols = combined_encoded.select_dtypes(include='bool').columns
combined_encoded[bool_cols] = combined_encoded[bool_cols].astype(int)

# ── 3. Split them back apart using the marker column ──
train_final = combined_encoded[combined_encoded['is_train']==1].drop(columns=['is_train'])
test_final = combined_encoded[combined_encoded['is_train']==0].drop(columns=['is_train','SalePrice'])

print("Train final shape:", train_final.shape)   # (1458, 244)
print("Test final shape:", test_final.shape)      # (1459, 243) — one less, no SalePrice column, correct

# ── 4. Train Ridge on the FULL train set (not just 80%, since we're done validating) ──
X = train_final.drop(columns=['SalePrice', 'Id'])
y = np.log1p(train_final['SalePrice'])

ridge = Ridge(alpha=10)
ridge.fit(X, y)

# ── 5. Predict on test, convert back from log scale ──
X_test = test_final.drop(columns=['Id'])
X_test = X_test[X.columns]   # ensures column order matches exactly what the model was trained on

test_preds_log = ridge.predict(X_test)
test_preds = np.expm1(test_preds_log)   # undo the log1p transform — back to real dollars

# ── 6. Build the submission file in Kaggle's expected format ──
submission = pd.DataFrame({'Id': test_final['Id'], 'SalePrice': test_preds})
submission.to_csv('submission.csv', index=False)
print(submission.head())
```
